```python
import numpy as np
import sympy as sp

x, y = sp.symbols('x y')
sp.init_printing()
np.random.seed(42)
print('Setup complete. SymPy', sp.__version__, '| NumPy', np.__version__)
```

```python
def f(x):
    return x**2
h=2
slope=f((4+h)-f(4))/h
print('Numerical f\'(4):', round(slope, 4), ' (exact = 6)')
```

```python
expr = x**2
deriv = sp.diff(expr, x)
print('d/dx (x**2) =', deriv)
print('Symbolic f\'(3) =', deriv.subs(x, 3))
```

```python
def g(x):
    return x**3 + 2*x
h=1e-6
slope2=(g(2+h)-g(2))/h
print("numerical dervivative",round(slope2,2))
der=sp.diff(g(x),x)
print('d/dx',der)
print("symbolic",der.subs(x,2))
```

```python
f2 = x**2 + 3*x*y + y**2
print('df/dx =', sp.diff(f2, x))
print('df/dy =', sp.diff(f2, y))
```
